```python
import cppimport
bindings = cppimport.imp("bindings")
```

```python
s = bindings.Sounding("example", False, False, 100)

print("SB Parcel Buoyancy Trace: ", s.get_sb_parcel_buoyancy())
print("ML Parcel Buoyancy Trace: ", s.get_ml_parcel_buoyancy())
print("MU Parcel Buoyancy Trace: ", s.get_mu_parcel_buoyancy())
print("SB Parcel EBuoyancy Trace: ", s.get_sb_entrainment_parcel_buoyancy())
print("ML Parcel EBuoyancy Trace: ", s.get_ml_entrainment_parcel_buoyancy())
print("MU Parcel EBuoyancy Trace: ", s.get_mu_entrainment_parcel_buoyancy())

print("SB Parcel Temp Trace: ", s.get_sb_parcel_temp())
print("ML Parcel Temp Trace: ", s.get_ml_parcel_temp())
print("MU Parcel Temp Trace: ", s.get_mu_parcel_temp())
print("SB Parcel E-Temp Trace: ", s.get_sb_entrainment_parcel_temp())
print("ML Parcel E-Temp Trace: ", s.get_ml_entrainment_parcel_temp())
print("MU Parcel E-Temp Trace: ", s.get_mu_entrainment_parcel_temp())

print("0-500m RH: ", s.get_rh(0, 500))
print("0-500m W: ", s.get_w(0, 500))
print("0-500m SRW: ", s.get_srw(0, 500))
print("0-500m SRH: ", s.get_srh(0, 500))
print("0-500m Shear: ", s.get_shear(0, 500))
print("0-500m Streamwise Vorticity: ", s.get_streamwise_vorticity(0, 500))

print("SBCAPE: ", s.get_sbcape())
print("MLCAPE: ", s.get_mlcape())
print("MUCAPE: ", s.get_mucape())
print("SBECAPE: ", s.get_sbecape())
print("MLECAPE: ", s.get_mlecape())
print("MUECAPE: ", s.get_muecape())

print("STP: ", s.get_stp())
```
