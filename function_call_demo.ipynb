```python
import cppimport
bindings = cppimport.imp("bindings")
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
s = bindings.Sounding("example", True, True, 100)
```

```python
mu_t_trace = np.ma.asarray(s.get_mu_parcel_temp(True))
mu_p_trace= np.ma.asarray(s.get_pressure(True))
```

```python
mu_p_trace
```

```python
delta_T = np.diff(mu_t_trace)

pressure_midpoints = (mu_p_trace[:-1] + mu_p_trace[1:]) / 2

plt.figure(figsize=(8, 5))
plt.plot(pressure_midpoints, delta_T, marker="o", linestyle="-", linewidth=2,)

plt.xlabel("Pressure")
plt.ylabel("Delta T")
plt.title("Pressure vs Delta T")

plt.grid(True, linestyle="--", alpha=0.6)

# Atmospheric profiles often invert or reverse the pressure axis (high pressure at bottom)
# Uncomment the following line if you want higher pressure on the left:
# plt.gca().invert_xaxis()

plt.tight_layout()
plt.show()
```
